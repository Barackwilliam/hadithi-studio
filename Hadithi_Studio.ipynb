# 🌐 Weka ukurasa wa kudumu kwenye Hugging Face (mara moja tu, bure)
Ukurasa utakuwa hewani kila wakati, kwa link ya kudumu, bila kufungua Colab. Picha zake zinatoka kwenye huduma ya bure ya mtandaoni (bila GPU).

**Kabla ya kuanza:**
1. Fungua akaunti ya bure kwenye [huggingface.co/join](https://huggingface.co/join).
2. Tengeneza token kwenye [huggingface.co/settings/tokens](https://huggingface.co/settings/tokens): **Create new token → aina "Write"**, kisha nakili.
3. Kwenye Colab, bonyeza 🔑 (**Secrets**) upande wa kushoto, kisha ongeza:
   - `HF_TOKEN` = token yako
   - `APP_PASSWORD` = neno la siri la kuingia kwenye ukurasa
   - `GEMINI_API_KEY` = *(si lazima)* key ya Gemini kwa ajili ya Chat

   Washa **Notebook access** kwa kila moja.

Kisha bonyeza ▶️ hapa chini. Unahitaji tu kuwa umeendesha Hatua ya 1 kwanza; GPU haihitajiki. Ukibadilisha code baadaye, bonyeza tena ili kusasisha ukurasa.

````python
#@title ▶️ Weka / sasisha ukurasa wa kudumu
APP_USER = "jamiitek"  #@param {type:"string"}
JINA_LA_SPACE = "hadithi-studio"  #@param {type:"string"}

import os, subprocess
from google.colab import userdata

def siri(jina):
    try:
        return (userdata.get(jina) or "").strip()
    except Exception as e:
        print(f"⚠️ Siri '{jina}' haipatikani ({type(e).__name__}). Iongeze kwenye 🔑 Secrets na uwashe 'Notebook access'.")
        return ""

HF_TOKEN, APP_PASSWORD = siri("HF_TOKEN"), siri("APP_PASSWORD")
if not os.path.exists("/content/hadithi-studio/space/tuma.py"):
    print("⚠️ Endesha Hatua ya 1 kwanza.")
elif HF_TOKEN and APP_PASSWORD:
    from huggingface_hub import HfApi
    try:
        jina_la_hf = HfApi(token=HF_TOKEN).whoami()["name"]
    except Exception as e:
        raise SystemExit(f"❌ HF_TOKEN haikubaliwi na Hugging Face. Tengeneza token mpya ya aina 'Write'.\n({e})")
    mazingira = dict(os.environ, HF_TOKEN=HF_TOKEN, APP_USER=APP_USER, APP_PASSWORD=APP_PASSWORD,
                     GEMINI_API_KEY=siri("GEMINI_API_KEY"), HF_SPACE=f"{jina_la_hf}/{JINA_LA_SPACE}")
    r = subprocess.run(["python", "/content/hadithi-studio/space/tuma.py"], env=mazingira, capture_output=True, text=True)
    print(r.stdout)
    if r.returncode != 0:
        print("❌ Kutuma kumeshindwa. Nakili ujumbe huu umpe msaidizi wako:\n")
        print(r.stderr[-3000:])
    else:
        print("⏳ Hugging Face inajenga ukurasa (dakika 5 hadi 10). Fungua link hapo juu, kisha ingia kwa:")
        print(f"   jina: {APP_USER}   |   neno la siri: (lile ulilohifadhi kwenye Secrets)")
````

## A. Andika hadithi yako ✍️
Badilisha maandishi yaliyo hapa chini, kisha bonyeza ▶️. Kanuni muhimu:
- **`picha:`** eleza tukio kwa **Kiingereza** (AI ya picha inaelewa Kiingereza vizuri zaidi).
- **`mazungumzo:`** andika kwa **Kiswahili**. Mstari usio na jina ni wa msimulizi. `- neema: "..."` maana yake Neema anaongea.
- **`wahusika:`** kwenye picha za tukio, taja wanaoonekana ili sura zao zifanane kila tukio.
- Hakikisha nafasi (indentation) zinabaki kama zilivyo.

💡 *Huna muda wa kuandika?* Tumia ChatGPT, Gemini au Claude bure kukuandikia hadithi kwa muundo huu. Angalia maelekezo kwenye README.

````python
%%writefile {HADITHI}
# ============================================================
#  HADITHI YA MFANO — nakili faili hili na ubadilishe upendavyo
# ============================================================

kichwa: "Siri ya Kisima cha Kijiji"

# Mtindo wa picha zote (Kiingereza hufanya kazi vizuri zaidi). Mifano:
#   "colorful 2D cartoon, children's storybook illustration"
#   "3D animated movie style, pixar-like, soft lighting"
#   "anime style, studio ghibli inspired, warm colors"
mtindo: "colorful 2D cartoon illustration, African village setting, warm sunset colors, storybook style"

# 16:9 = YouTube | 9:16 = TikTok/Reels/Shorts | 1:1 = Instagram
ukubwa: "16:9"

# ---------- WAHUSIKA ----------
# sauti: rehema (mwanamke TZ), daudi (mwanamume TZ), zuri (mwanamke KE), rafiki (mwanamume KE)
# kina: "+25Hz" = sauti nyembamba (mtoto), "-15Hz" = sauti nzito (mzee)
# kasi: "-10%" = polepole, "+10%" = haraka
wahusika:
  msimulizi:
    sauti: daudi
    kasi: "-5%"

  amani:
    jina: Amani
    maelezo: "Amani, a 10 year old Tanzanian boy, short black hair, red t-shirt, blue shorts, big curious eyes"
    sauti: rafiki
    kina: "+25Hz"
    kasi: "+5%"

  neema:
    jina: Neema
    maelezo: "Neema, a 9 year old Tanzanian girl, braided hair with yellow beads, yellow dress"
    sauti: rehema
    kina: "+20Hz"

  bibi:
    jina: Bibi Zawadi
    maelezo: "Bibi Zawadi, a kind elderly African grandmother, grey hair, colorful kanga cloth, walking stick"
    sauti: zuri
    kina: "-10Hz"
    kasi: "-12%"

# ---------- MATUKIO ----------
# picha:     maelezo ya picha (Kiingereza ni bora)
# wahusika:  nani anaonekana kwenye picha (husaidia sura zifanane)
# mazungumzo: mstari bila jina = msimulizi;  "- amani: ..." = Amani anaongea
# mwendo (si lazima): karibia, mbali, kulia, kushoto, tuli
matukio:
  - picha: "wide view of a small African village at sunrise, mud houses, baobab tree, an old stone well in the center"
    mwendo: karibia
    mazungumzo:
      - "Hapo zamani za kale, katika kijiji kidogo cha Mbuyuni, palikuwa na kisima cha ajabu."
      - "Hakuna aliyejua siri yake, mpaka siku moja..."

  - picha: "a boy and a girl running happily on a dirt path towards an old stone well"
    wahusika: [amani, neema]
    mazungumzo:
      - amani: "Neema! Njoo haraka! Nimesikia sauti ikitoka ndani ya kisima!"
      - neema: "Amani, acha utani wako. Visima haviongei!"

  - picha: "a boy and a girl leaning over a stone well, looking down with surprised faces, magical blue glow coming from the well"
    wahusika: [amani, neema]
    mwendo: karibia
    mazungumzo:
      - "Walipochungulia ndani, waliona mwanga wa buluu ukimeta kama nyota."
      - neema: "Mungu wangu! Unaona kile ninachokiona?"

  - picha: "an elderly grandmother with a walking stick walking towards the well, smiling wisely"
    wahusika: [bibi]
    mwendo: kulia
    mazungumzo:
      - bibi: "Watoto wangu, msiogope. Kisima hiki kinang'aa kwa wale wenye mioyo safi tu."
      - amani: "Bibi, wewe unaijua siri yake?"

  - picha: "grandmother sitting under a baobab tree telling a story to a boy and a girl, warm evening light"
    wahusika: [bibi, amani, neema]
    mwendo: mbali
    mazungumzo:
      - bibi: "Zamani, wanakijiji walishirikiana kuchimba kisima hiki wakati wa ukame mkubwa."
      - bibi: "Tangu siku hiyo, maji yake hayajawahi kukauka, kwa sababu umoja wao uliishi ndani yake."

  - picha: "the whole village celebrating around the glowing well at night, people dancing, lanterns, happy children"
    wahusika: [amani, neema]
    mwendo: kushoto
    kimya: 1.5
    mazungumzo:
      - neema: "Kwa hiyo siri ya kisima ni umoja wetu!"
      - "Na tangu siku hiyo, Amani na Neema walikuwa walinzi wa kisima cha Mbuyuni."
      - "Mwisho."
````
